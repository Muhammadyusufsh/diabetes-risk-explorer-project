```python
# Import the tools we will use
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from ucimlrepo import fetch_ucirepo

pd.set_option("display.max_columns", 30)
sns.set_theme(style="whitegrid")
```

```python
# Download CDC Diabetes Health Indicators directly from UCI
# Dataset ID 891: https://archive.ics.uci.edu/dataset/891/cdc+diabetes+health+indicators
dataset = fetch_ucirepo(id=891)

X = dataset.data.features.copy()      # the input columns
y = dataset.data.targets.copy()       # what we want to classify
df = pd.concat([X, y], axis=1)

print(f"Rows: {df.shape[0]:,}")
print(f"Columns: {df.shape[1]}")
df.head()
```

```python
# First inspection: column names, data types, and missing values
print("Column names:")
print(df.columns.tolist())

print("\nMissing values per column:")
print(df.isna().sum().sort_values(ascending=False).head(10))

print("\nTarget distribution:")
print(df["Diabetes_binary"].value_counts().rename({0: "No reported diabetes", 1: "Prediabetes or diabetes"}))
```

```python
# Your first chart: outcome balance
ax = sns.countplot(data=df, x="Diabetes_binary", hue="Diabetes_binary", legend=False, palette="Set2")
ax.set_xticks([0, 1], ["No reported diabetes", "Prediabetes / diabetes"])
ax.set_xlabel("Self-reported outcome")
ax.set_ylabel("Number of survey respondents")
ax.set_title("Outcome distribution in the CDC survey")
plt.show()
```
